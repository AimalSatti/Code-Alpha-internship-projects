- # DATA EXPLORATION.

```python
# Import libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# load the dataset

df = pd.read_csv('../datasets/Advertising.csv')
```

```python
# let's explore the dataset

print('                                         TOP 5 ROWS OF THE DATASET\n                              ')
print(df.head())    
print('\n                                         DATASET INFORMATION\n                              ')
print(df.info())
print('\n                                         DATASET DESCRIPTION \n                              ')
print(df.describe())
print('\n                                         DATASET SHAPE\n                               ')
print(df.shape)
print('\n                                         DATASET COLUMNS\n                               ')
print(df.columns)
```

# key insights from the above information are:
1. The dataset has 200 rows and 5 columns.
2. The columns are: Unnamed, TV, Radio, Newspaper, and Sales.
3. The Sales column is the target variable, and the other four columns are the features.
4. The Sales column is a continuous variable, and the other three columns are also continuous variables .

```python
# let's drop the unnecessary column 'Unnamed: 0' from the dataset

df.drop(columns=['Unnamed: 0'], inplace=True)
```

```python
# let's check the missing values and duplicates in the dataset

print('\n                                         MISSING VALUES\n                               ')
print(df.isnull().sum())
print('\n                                         DUPLICATES\n                               ')
print(df.duplicated().sum())
```

# key insights from the above information are:
1. The dataset does not contain any missing values.
2. The dataset does not contain any duplicate rows.
3. Therefore, we can conclude that the dataset is clean and ready for analysis.

# *DATA VISUALIZATION.*

```python
# let's check for any outliers in the dataset using boxplots.

num_cols = ['TV', 'Radio', 'Newspaper']
plt.figure(figsize=(12, 4))
for i, col in enumerate(num_cols, 1):
    plt.subplot(1, 3, i)
    sns.boxplot(y=df[col], color='skyblue')
    plt.title(f'Boxplot of {col}')

plt.tight_layout()
plt.show()
```

```python
# let's see whether the features are normally distributed or not?

print(sns.histplot(df['TV'], kde=True))


```

```python
print(sns.histplot(df['Radio'], kde=True))
```

```python
print(sns.histplot(df['Newspaper'], kde=True))
```
